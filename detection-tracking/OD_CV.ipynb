```python
import cv2
```

```python
cap = cv2.VideoCapture(0)

# loop runs if capturing has been initialized.
MOGObject = cv2.createBackgroundSubtractorMOG2() 

while True:
    # reads frames from a video
    ret, org_frame = cap.read()
      
    # convert to gray scale of each frames
    gray = cv2.cvtColor(org_frame, cv2.COLOR_BGR2GRAY)
    
    # Background_subraction to find the moiving object
    Background_subraction = MOGObject.apply(gray)
    
    #blur and threshold the image image thershold 
    Gaussian = cv2.GaussianBlur(Background_subraction, (7, 7), 0)
    
    ret, thresh = cv2.threshold(Gaussian, 120, 255, cv2.THRESH_BINARY)
    
    # detect contours using blue channel
    contours , hi = cv2.findContours(thresh, cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)# finding contours
    for c in contours:
        # get bounding rect
        (x,y,w,h) = cv2.boundingRect(c)
        # draw red rect
        cv2.rectangle(org_frame, (x,y), (x+w,y+h), (0, 0, 255), 2)    
        # count of discovered contours        
    
#     #Display frames in a window 
    cv2.imshow('Background subraction', org_frame)
      
    # Wait for Esc key to stop
    k = cv2.waitKey(30) & 0xFF
    if k == 27:
        break
  
 #De-allocate any associated memory usage
cv2.destroyAllWindows()
```
